```python
import requests
import json
import pandas as pd
import time

MODEL_NAME = "qwen2.5:0.5b"
OLLAMA_URL = "http://localhost:11434/api/generate"

!ollama pull {MODEL_NAME}

def myllm(user_text):
    my_data = {
        "model": MODEL_NAME,
        "prompt": user_text,
        "stream": False
    }

    try:
        response = requests.post(OLLAMA_URL, json=my_data)

        if response.status_code == 200:
            return response.json().get('response', 'Пустой ответ')
        else:
            return f"Ошибка сервера: {response.status_code}"

    except Exception as err:
        return f"Что-то пошло не так: {str(err)}"

my_prompts = [
    "напиши мне привет на любых 3 языках",
    "напиши самый простой код на Python, чтобы вывести 'Hello World'",
    "как объяснить бабушке, что такое VPN?",
    "как зовут тони старка?",
    "в чем разница между списком (list) и кортежем (tuple) в питоне?",
    "посоветуй, какой язык учить после Python для бэкенда, я выбираю между английским и испанским.",
    "что такое токенизация в анализе текста? Объясни по-простому для ребенка",
    "ах, если бы сбылась моя мечта, в каком мультике такая песня?",
    "когда вышел elden ring?",
    "смерть короля лича, что он сказал отцу?"
]

final_results = []

for i, text in enumerate(my_prompts, 1):
    print(f"Обрабатываю вопрос №{i}")

    llm_answer = myllm(text)

    final_results.append({
        "Запрос к LLM": text,
        "Вывод LLM": llm_answer.strip()
    })

report_table = pd.DataFrame(final_results)

print("Отчет")
display(report_table)

report_table.to_csv("my_lab2_report.csv", index=False)
```
